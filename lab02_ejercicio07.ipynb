```python
def minimax(piedras, es_maximizador):
    if piedras == 0:
        return (1 if es_maximizador else -1, None)

    if es_maximizador:
        mejor_valor = -float('inf')
        mejor_movimiento = 1
        for movimiento in [1, 2, 3]:
            if piedras - movimiento >= 0:
                valor, _ = minimax(piedras - movimiento, False)
                if valor > mejor_valor:
                    mejor_valor = valor
                    mejor_movimiento = movimiento
        return mejor_valor, mejor_movimiento
    else:
        mejor_valor = float('inf')
        mejor_movimiento = 1
        for movimiento in [1, 2, 3]:
            if piedras - movimiento >= 0:
                valor, _ = minimax(piedras - movimiento, True)
                if valor < mejor_valor:
                    mejor_valor = valor
                    mejor_movimiento = movimiento
        return mejor_valor, mejor_movimiento

def jugar_nim():
    piedras = 10
    turno_actual = False

    print(f"¡Comienza el juego de NIM! Piedras iniciales: {piedras}")
    print("Reglas: Quita 1, 2 o 3 piedras. Pierde el que se queda sin piedras.\n")

    while piedras > 0:
        print(f"Piedras restantes en la pila: {piedras}")

        if not turno_actual:
            movimiento = 0
            while movimiento not in [1, 2, 3] or movimiento > piedras:
                try:
                    movimiento = int(input("Tu turno (Min) - ¿Cuántas piedras quieres quitar (1, 2 o 3)?: "))
                except ValueError:
                    movimiento = 0

            piedras -= movimiento
            if piedras == 0:
                print("\n¡Te has quedado sin piedras! Gana la Computadora (MAX).")
                break
            turno_actual = True
        else:
            print("Turno de la Computadora (Max)...")
            _, movimiento = minimax(piedras, True)

            if movimiento > piedras:
                movimiento = piedras

            print(f"La computadora quita {movimiento} piedra(s).")
            piedras -= movimiento

            if piedras == 0:
                print("\n¡La computadora se ha quedado sin piedras! ¡Ganas tú (MIN)!")
                break
            turno_actual = False

        print("-" * 40)

if __name__ == "__main__":
    jugar_nim()
```
