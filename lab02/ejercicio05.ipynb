```python
palabras_a = ["sol", "computadora", "gato", "elefante"]
res_a = list(filter(lambda x: len(x) > 4, palabras_a))

lista_b = [1, None, "hola", None, 5]
res_b = list(filter(lambda x: x is not None, lista_b))

palabras_c = ["avion", "casa", "oso", "perro", "iglesia"]
res_c = list(filter(lambda x: x.lower().startswith(('a', 'e', 'i', 'o', 'u')), palabras_c))

palabras_d = ["radar", "casa", "reconoce", "python"]
res_d = list(filter(lambda x: x == x[::-1], palabras_d))

numeros = [12, 15, 23, 35, 40, 55]
res_e = list(filter(lambda x: x % 10 == 5, numeros))

productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]
res_f = list(filter(lambda p: p["precio"] > 100, productos))
```
